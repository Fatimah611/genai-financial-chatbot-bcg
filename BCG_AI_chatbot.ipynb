```python
import pandas as pd

data = pd.read_excel(r"DATA EXTRACTION.xlsx")

latest = data.iloc[-1]   # last row (sab se recent year)
prev   = data.iloc[-2]   # us se pehla year
```

```python
def simple_chatbot(user_query: str) -> str:
    q = user_query.lower().strip()

    if "total revenue" in q:
        return f"Total revenue is {latest['Total Revenue']:,}."
    elif "net income" in q and "last year" in q:
        change = latest['Net Income'] - prev['Net Income']
        direction = "increased" if change > 0 else "decreased"
        return f"Net income has {direction} by {abs(change):,} over the last year."
    elif "total assets" in q:
        return f"Total assets are {latest['Total Assets']:,}."
    else:
        return "Sorry, I can only answer predefined queries."
```

```python
if __name__ == "__main__":
    print("Financial Chatbot – type 'exit' to quit\n")
    while True:
        user = input("You: ")
        if user.lower() == "exit":
            print("Bot: Goodbye!")
            break
        print("Bot:", simple_chatbot(user))
exit
```
